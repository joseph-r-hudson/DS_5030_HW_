## Expected value of any gamble in European roulette

For each bet in European roulette, you can gain (36/K)-1 for your bet, where K is the number of spots you bet on. You can also lose whatever you bet on and in this instance we will say that is -1. So, betting a dollar has a chance for a person to gain 36/K-1 or lose 1. The probability of hitting any bet in European roulette is k/37, as there is 37 total spots. The probability of not hitting your bet is (37-k)/37, as this covers the probability of the spaces you don't bet on. To find the expectation, we just need to multiply the probability of each outcome by that outcome and take the sum. ## Expected value of rolling a die, where you continue to roll the die if you get a six and add each successive role to your recorded value

```python
values=[]

def roll_die():
    value=(random.randint(1,6))
    if value==6:
        value=value+roll_die()
    if value <6:
        value=value
    return value

for i in range(1000000):
    values.append(roll_die())
    
pd.Series(values).mean()

```

In the code above I wrote a function to simulate rolling a die, and continuing rolling the die if a die roll is a 6 and adding any successive values to the total. I simulated the function a millions time, and got a mean value around 4.2

## Showing that if $X$ is a normally distributed random variable, then $a + bX$ is distributed normally with mean $a+ b \mathbb{E}[X]$ and variance $b^2 \sigma_X^2$ 

```python
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt 


sample = np.random.normal(0,1,100000)

Ev_of_sample = np.mean(sample)
print(f'Expected Value of X: {Ev_of_sample}')

v_of_EX = np.var(sample)
print(f'Variance of X: {v_of_EX}')
```

Above I simulated a 100000 size sample for a normally distributed random variable from the normal distribution with mean 0 and standard deviation 1. The simluated sample has a mean very close to 0, and a variance very close to 1.

```python
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt 

a=2
b=3

sample = a+b*np.random.normal(0,1,100000)

Ev_of_sample = np.mean(sample)
print(f'Expected Value of X: {Ev_of_sample}')

v_of_EX = np.var(sample)
print(f'Variance of X: {v_of_EX}')
```

In the code above, I took my simulation for a random variable sample from the normal disbitrubiton with mean 0 and standard deviation 1, and I added a to the product of b and the sample. I chose a to be 2 and b to be 3. So, now the sample is a+bX, where X is a sample from the normal distribution and a is a constant 2 and b is a constant 3. As seen above, the expected value, or mean, of the sample now is 2, which is equal a+bX which is 2+(3 *0 ) = 2. The variance of a+bX sample is 9, which is equal to b^2 *V[X] which is (3^2) *1 = 9. So, the simulation shows that for a normally distributed random  variable X, a+bX is normally distirubted with mean a+bE[X] and variance of b^2*V[X]
