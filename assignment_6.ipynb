You can use https://docs.scipy.org/doc/scipy/reference/stats.html to generate values for the pdf/cdf and generate samples of random variates.


- Logistic distribution (similar to normal): 
$$
F(x; \sigma) = \dfrac{1}{1+e^{-x/\sigma}} 
$$ 
with $\sigma >0$.
- Exponential distribution (similar to log-normal):
$$
F_X(x) = \begin{cases}
0, & x<0 \\
1 - e^{-\lambda x}, & x \ge 0,
\end{cases}
$$
with $\lambda > 0$.
- Negative Binomial (similar to Poisson): The probability mass function for positive integers is:
$$
f(k;r,p) = \dfrac{(k+r-1)!}{k!(r-1)!}(1-p)^k p^r, \quad \text{ for $k=0,...n$}
$$
You can interpret this as follows: Flip a coin that comes up heads with probability $p$ until you get $r$ heads, and then stop. What is the probability of stopping at each $k=0,1,2,...$?
- Categorical (similar to Bernoulli): The probability mass function over $k = 1, 2, ..., K$ categories is
$$
f(k;p_1,...,p_K) = p_1^{k=1}p_2^{k=2}...p_K^{k=K}
$$
where $0 \le p_i \le 1$ and $ \sum_{k=1}^K p_k = 1$.

### Density and Distribution for variety of parameter values

```python
params = [(0, 1), (0, 2), (2, 1)]

# --- Define x-range ---
x = np.linspace(-10, 10, 400)

# --- Plot PDF and CDF ---
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

for mu, s in params:
    pdf = logistic.pdf(x, loc=mu, scale=s)
    cdf = logistic.cdf(x, loc=mu, scale=s)
    ax[0].plot(x, pdf, label=f'μ={mu}, s={s}')
    ax[1].plot(x, cdf, label=f'μ={mu}, s={s}')

ax[0].set_title('Logistic PDF')
ax[0].set_xlabel('x')
ax[0].set_ylabel('Density')

ax[1].set_title('Logistic CDF')
ax[1].set_xlabel('x')
ax[1].set_ylabel('Probability')

for a in ax:
    a.legend()
    a.grid(True)

plt.tight_layout()
plt.show()
```

### Comparison of KDE and ECDF with theoretical PDF and CDF

```python
sample= logistic.rvs(size=1000)
```

```python
sns.kdeplot(data=sample).set(title='KDE of Sample Logistic Distribution Data')
```

```python

x = np.linspace(-8,8, 100)

fig, ax = plt.subplots(figsize=(12, 8))

ax.plot(x, logistic.pdf(x), 'r-', lw=5, alpha=0.6, label='logistic pdf')


ax.set_title('Theoretical Logistic Distribution PDF')
ax.set_xlabel('x')
ax.set_ylabel('Density')

plt.show()
```

We can see that the KDE and theoretical PDF plot for the logistic distribution are quite similar.

```python
#ECDF plot
sns.ecdfplot(data=sample).set(title='ECDF of Sample Logistic Distribution Data')
```

```python
fig, ax = plt.subplots(figsize=(12, 8))

ax.plot(x, logistic.cdf(x), 'r-', lw=5, alpha=0.6, label='logistic cdf')


ax.set_title('Theoretical Logistic Distribution CDF')
ax.set_xlabel('x')
ax.set_ylabel('Density')

plt.show()
```

Similar to the KDE and PDF plots, the ECDF and CDF plots of the logistic distribution are quite similar.
