- Logistic distribution (similar to normal): 
$$
F(x; \sigma) = \dfrac{1}{1+e^{-x/\sigma}} 
$$ 
with $\sigma >0$.
- Exponential distribution (similar to log-normal):
$$
F_X(x) = \begin{cases}
0, & x<0 \\
1 - e^{-\lambda x}, & x \ge 0,
\end{cases}
$$
with $\lambda > 0$.
- Negative Binomial (similar to Poisson): The probability mass function for positive integers is:
$$
f(k;r,p) = \dfrac{(k+r-1)!}{k!(r-1)!}(1-p)^k p^r, \quad \text{ for $k=0,...n$}
$$
You can interpret this as follows: Flip a coin that comes up heads with probability $p$ until you get $r$ heads, and then stop. What is the probability of stopping at each $k=0,1,2,...$?
- Categorical (similar to Bernoulli): The probability mass function over $k = 1, 2, ..., K$ categories is
$$
f(k;p_1,...,p_K) = p_1^{k=1}p_2^{k=2}...p_K^{k=K}
$$
where $0 \le p_i \le 1$ and $ \sum_{k=1}^K p_k = 1$.

### Density and Distribution for variety of parameter values

```python
# NOTE:  I got this code from asking chat gpt to plot the pdf and cdf for different parameters of the logistic distribution

params = [(0, 1), (0, 2), (2, 1), (2, 2), (3, 1)]

# --- Define x-range ---
x = np.linspace(-10, 10, 400)

# --- Plot PDF and CDF ---
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

for mu, s in params:
    pdf = logistic.pdf(x, loc=mu, scale=s)
    cdf = logistic.cdf(x, loc=mu, scale=s)
    ax[0].plot(x, pdf, label=f'μ={mu}, s={s}')
    ax[1].plot(x, cdf, label=f'μ={mu}, s={s}')

ax[0].set_title('Logistic PDF')
ax[0].set_xlabel('x')
ax[0].set_ylabel('Density')

ax[1].set_title('Logistic CDF')
ax[1].set_xlabel('x')
ax[1].set_ylabel('Probability')

for a in ax:
    a.legend()
    a.grid(True)

plt.tight_layout()
plt.show()
```

The support of the negative binomial distribution is (0,$\infty$), because its models the amount of trials needed to reach a certain number of successes, like how many flips of a coin needed to get a certain amount of heads.

### Density and Distribution for variety of parameter values

```python
params = [(1, 0.5), (1, 0.25), (2, .7), (3, .4), (4, .5)]

# --- Define x-range ---

x = np.arange(0, 20) 

# --- Plot PDF and CDF ---
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

for n, p in params:
    pdf = nbinom.pmf(x, n=n, p=p)
    cdf = nbinom.cdf(x, n=n, p=p)
    ax[0].plot(x, pdf, label=f'n={n}, p={p}')
    ax[1].plot(x, cdf, label=f'n={n}, p={p}')

ax[0].set_title('Negative Binomial PMF')
ax[0].set_xlabel('x')
ax[0].set_ylabel('Density')

ax[1].set_title('Negative Binomial CDF')
ax[1].set_xlabel('x')
ax[1].set_ylabel('Probability')

for a in ax:
    a.legend()
    a.grid(True)

plt.tight_layout()
plt.show()
```

# Categorical Distribution

### Support

The support of a cateogircal distribution are the cateogires that serves as potential outcomes. Likethe rolls of the a die are a cateogircal distribution with the sides of the die serving as the cagtegories, and the probabilities of the die landing on each side as their corresponding probabilities. So the support depends on the possible categories that are outcomes.

### Density and Distribution for variety of parameter values

```python
categories  = ["A","B","C","D","E"]

params = [[0.1,0.2,0.3,0.2,0.2], [0.2,0.3,0.2,0.2,0.1], [0.4,0.1,0.2,0.2,0.1], [0.2,0.2,0.2,0.3,0.1], [0.5,0.1,0.1,0.1,0.3]]

for probs in params:
    plt.bar(categories, probs, color='blue')
    plt.title('Cateogoriral Distribution PMF')
    

    for i, prob in enumerate(probs):
        plt.text(i, prob, f'{prob:.2f}', ha='center', va='bottom')
    plt.show()
```

```python
# I got this code by asking chat gpt to make cdfs for the cateogrical distribution given my categories  = ["A","B","C","D","E"] and their differents probability parameters
for probs in params:

    cdf = np.cumsum(probs)
    
    # Plot as a step function
    plt.step(categories, cdf, where='mid')
    plt.ylim(0, 1.05)
    plt.ylabel('CDF')
    plt.title('Categorical Distribution CDF')
    plt.show()
```
